## 1st_Question

```python
import re
 
# Explanation
"""
First we will open the file that has the emails 
Then we will give the pattern to get all the mails
[A-Za-z0-9]{1,16} : It means that we put 1 to 16 lowercase or uppercase letter  or digit.

@[a-z]+.[a-z]{2,}: we put the @ then [a-z]+.[a-z]{2,} 
                   means that we will search for lowercase letters in the domain
                   .[a-z]{2,} means that after the dot there will be 2 or more lowercase letters
then we will use for loop to pass through each line 
and search for the mails that match the pattern.
""" 
file =open('emails.txt', 'r')
email_pattern = r'[A-Za-z0-9]{1,16}@[a-z]+.[a-z]{2,}'
#Iterate through each line in the file
for line in file:
    line = line.strip() 
    #Use regular expression to find email addresses after ":"
    match = re.search(r': *([A-Za-z0-9]+@[a-z]+.[a-z]{2,})', line)
    if match:
        email = match.group(1)
        print(email)
file.close()
```

## 2nd_Question

```python
"""
First we will open the file that has the Passwords
Then we will give the pattern to get all the valid Passwords
(?=.*[A-Z]) : It means that the password should have at least one uppercase letter.

(?=.*[a-z]{6,}): It means that the password should have 6 lowercase letters or more.

(?=.*[0-9]) : It means that the password should have at least one digit.

{8,} : It means that the password should have at least 8 of upper or lower or digit .

then we will use for loop to pass through each line 
and search for the mails that match the pattern.

"""

password_pattern = r'^(?=.*[A-Z])(?=.*[a-z]{6,})(?=.*[0-9]).{8,}'
file = open('passwords.txt', 'r')


for line in file:
    line = line.strip()  # Remove leading/trailing whitespace
    if re.match(password_pattern, line):
        print(line)
```

## 3rd_Question

```python

"""
First we will open the file that has the Passwords
Then we will give the pattern to get all the valid Visa OR Mastercard

VISA PATTERN:
^4[0-9]{3} : It means that Visa must start with 4 then followed with 3 digits.

( [0-9]{4}){2}) : It means that Visa after the first 4 digits should have another
                another two groups of 4 digits seperated by space.

( [0-9]{1,4}) : It means that the last group of digits will be a group from 1 to 4 digits.
________________________________________________________________________________________

MASTERCARD PATTERN:

^5[0-9]{3} : It means that Visa must start with 5 then followed with 3 digits.

( [0-9]{4}){3} : It means that Visa after the first 4 digits should have another
                another three groups of 4 digits seperated by space.
                
________________________________________________________________________________________

then we will use for loop to pass through each line 
and search for the VISA or MASTERCARD that match the pattern.

"""
visa_pattern = r'^4[0-9]{3}( [0-9]{4}){2}( [0-9]{1,4})'
mastercard_pattern = r'^5[0-9]{3}( [0-9]{4}){3}'

file = open('cards.txt', 'r')

# We will Iterate through each line in the file
for line in file:
    line = line.strip()
    if re.match(visa_pattern, line):
        print("Visa:", line)
    elif re.match(mastercard_pattern, line):
        print("Mastercard:", line)
file.close()
```

### Tuple

```python
# Create a tuple
t = ("apple", "banana", "cherry")

print("Our main Tuple:", t)
print("Access First element: ", t[0])


t = t + ("orange",)
print("Tuple after adding: ", t)


t = (element for element in t if element != "apple")


for item in t:
    print(item)
```

### Dictionary

```python

my_dict = {"apple": 3, "banana": 5, "cherry": 2}
print("Access Value for 'apple':", my_dict["apple"])
my_dict["orange"] = 4
print("Dictionary after adding 'orange':", my_dict)

my_dict["apple"] = 6
print("Dictionary after changing 'apple' value:", my_dict)
del my_dict["cherry"]
print("Dictionary after removing 'cherry':", my_dict)
for key, value in my_dict.items():
    print(f"Key: {key}, Value: {value}")
```

### Set

```python

s = {"apple", "banana", "cherry"}
print("Our main Set: ", s)
print("Access Elements:")
for item in s:
    print("Access Element: ", item)
print("\nAdding an Element:")
s.add("orange")
print("Set after adding: ", s)
print("\nChanging an Element:")
s.remove("apple")  
s.add("kiwi")       
print("Set after changing: ", s)

print("\nRemoving an Element:")
s.remove("kiwi")  
print("Set after removing: ", s)
print("\nLooping through the Set:")
for item in s:
    print("Element: ", item)
```
