```python
import tkinter as tk
import pandas as pd
import json
import random
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import LabelEncoder

# Load data from the intents.json file
data_path = "C:\\Users\\Lenovo\\Downloads\\archive(2)\\intents.json"
with open(data_path, 'r') as file:
    data = json.load(file)

# Prepare data for Naive Bayes model
intents = data['intents']
patterns = []
tags = []
responses = []

for intent in intents:
    for pattern in intent['patterns']:
        patterns.append(pattern)
        tags.append(intent['tag'])
        responses.append(intent['responses'])

df = pd.DataFrame({'patterns': patterns, 'tag': tags, 'responses': responses})

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['patterns'])
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(df['tag'])

# Train Naive Bayes model
nb_model = MultinomialNB()
nb_model.fit(X, y)

# GUI
def get_response():
    user_input = entry.get()
    user_input_transformed = vectorizer.transform([user_input])
    predicted_tag = nb_model.predict(user_input_transformed)[0]
    tag = label_encoder.inverse_transform([predicted_tag])[0]
    responses_list = df[df['tag'] == tag]['responses'].values[0]
    response = random.choice(responses_list)
    response_text.insert(tk.END, "Bot: " + response + "\n\n")

root = tk.Tk()
root.title("Mental Health Chatbot")

label = tk.Label(root, text="Enter your message:")
label.pack()

entry = tk.Entry(root, width=50)
entry.pack()

button = tk.Button(root, text="Send", command=get_response)
button.pack()

response_text = tk.Text(root, width=50, height=10)
response_text.pack()

root.mainloop()
```
