## Creating The Graph

```python
# Creating a Graph

import tensorflow as tf

# Creating nodes with constant NO. 
a=tf.constant(5)

b=tf.constant(2)

c=tf.constant(3)

# Creating nodes with Operations
d=tf.multiply(a,b)

e=tf.add(c,b)

f=tf.subtract(d,e)
```
