# Questions

## What is a perceptron?

A perceptron is a simple neural network model that is used for binary classification problems

## List the activation function used in perceptron and the differences between them?

The activation function used in perceptron is usually a step function or a sigmoid function. A step function outputs either 0 or 1, while a sigmoid function outputs a continuous value between 0 and 1


## How is the output of a perceptron computed?

By taking the sum of the weighted input values and passing it through an activation function

## What is the role of a loss function in the training of a perceptron?

The role of a loss function is to measure the difference between the predicted output and the actual output for an input.

### Define mean squared error (MSE) loss function.


used to calculate the average squared difference between the predicted output and the actual output for an input.


MSE = (1/n) * Σ(y - ŷ)²



## How can you adjust the weights of a perceptron during training?

The gradient descent approach can be used to modify a perceptron's weights as it is being trained. In order to minimize the loss function, the method calculates the gradient of the loss function with respect to the weights and then updates the weights in the opposite direction as the gradient.

## Why do we apply sign() function on the perceptron output?

The perceptron's continuous output is changed into a binary output of either 0 or 1 using the sign() function.

## What is the use of Bias?

Bias is used to shift the activation function to the left or right

## How to update Bias?

The gradient of the loss function with respect to the bias can be calculated, and the bias can then be changed in the same manner as the weights.

## Given the table below,  Consider a perceptron with weights w1=−0.5, w2=  0.1,  and w3= 0.4,  bias  b= 0.2,  and learning rate α= 0.01.  Train for 2 iteration with weights updated after each sample and then calculate the accuracy


```python
import numpy as np

# step function
def step(x):
    return np.where(x >= 0, 1, -1)

# weights and bias
w = np.array([-0.5, 0.1, 0.4])
b = 0.2

lr = 0.01
n_iter = 2

# the dataset
X = np.array([[0.2, 0.5, 0.7],
              [0.1, 0.3, 0.6],
              [0.4, 0.8, 0.2],
              [0.6, 0.4, 0.9]])
y = np.array([1, -1, 1, -1])

for i in range(n_iter):
    for j in range(len(X)):

        z = np.dot(X[j], w) + b
        y_pred = step(z)
        
        error = y[j] - y_pred
        
        w += lr * error * X[j]
        b += lr * error

y_pred = step(np.dot(X, w) + b)
accuracy = np.mean(y_pred == y)
print("Accuracy:", accuracy)
print("Weights:", w)
print("Bias:", b)
```

## Write code to repeat question 11 but with the weights updated after a full iteration using MSE loss, compare the final weights and accuracy.

```python
# Define hyperparameters
learning_rate = 0.01
epochs = 2

# Define Mean Squared Error (MSE) loss function
def mse_loss(y_true, y_pred):
    return np.mean(np.square(y_true - y_pred))


for epoch in range(epochs):
    shuffle_indices = np.random.permutation(len(X))
    X = X[shuffle_indices]
    y = y[shuffle_indices]
    
    dw = np.zeros_like(w)
    db = 0
    
    for i in range(len(X)):
        z = np.dot(w, X[i]) + b
        y_pred = np.sign(z)
        
        error = y[i] - y_pred
        
        dw += learning_rate * error * X[i]
        db += learning_rate * error
        
    w += dw
    b += db
    
    y_pred = np.sign(np.dot(X, w) + b)
    accuracy = np.mean(y_pred == y)
    
    print(f"Epoch {epoch+1} -- Loss: {mse_loss(y, y_pred):.4f} -- Accuracy: {accuracy:.4f}")

print(f"\nFinal weights: {w}")
print(f"Final bias: {b}")
```
