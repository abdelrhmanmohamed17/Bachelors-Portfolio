### 6.  Having matrixA and matrixB both with shape (3,2) how to create a numpy array withshape (2,3,2) using np.concatenate?

```python
import numpy as np
matrixA= np.random.randint(100, size=(3,2))
matrixB= np.random.randint(100, size=(3,2))
matrixC=np.concatenate((matrixA, matrixB), axis=0, out=None, dtype=None, casting="same_kind")
print("The new matrix will be:\n",matrixC)
```
