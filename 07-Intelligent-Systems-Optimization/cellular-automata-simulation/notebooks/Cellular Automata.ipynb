```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def generate_CA(rule, size=30, steps=30):
   
    ca = np.zeros((steps, size), dtype=int)
    ca[0, size // 2] = 1  
    for i in range(1, steps):
        for j in range(1, size - 1):
            ca[i, j] = rule[int("".join(map(str, ca[i - 1, j - 1:j + 2])), 2)]

    return ca
rule = np.array([0, 1, 1, 0, 1, 1, 0, 0], dtype=int)
```

```python

def plot_CA(ca):
    plt.imshow(ca, cmap='Blues', interpolation='nearest')
    plt.xlabel('Cell')
    plt.ylabel('Time Step')
    plt.title('1D Cellular Automaton')
    plt.show()

```

```python
automaton = generate_CA(rule)
plot_CA(automaton)
```
