```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import cv2
import numpy as np
```

Loading Cifar 10 dataset

```python
# Load the CIFAR-10 dataset
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()
# Preprocess the data using OpenCV
def preprocess_images(images):
 processed_images = []
 for img in images:
    # Resize the image to a specific size (e.g., 64x64)
    img = cv2.resize(img, (64, 64))
 # Normalize pixel values to be between 0 and 1
     
    processed_images.append(img)
 return np.array(processed_images)
train_images = preprocess_images(train_images)
test_images = preprocess_images(test_images)
```

```python
# Define the CNN model
model1 = models.Sequential()
# Convolutional layers for feature extraction
model1.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 3)))
model1.add(layers.MaxPooling2D((2, 2)))
model1.add(layers.Conv2D(64, (3, 3), activation='relu'))
model1.add(layers.MaxPooling2D((2, 2)))
model1.add(layers.Conv2D(64, (3, 3), activation='relu'))
# Flatten the output for fully connected layers
model1.add(layers.Flatten())
# Fully connected layers for classification
model1.add(layers.Dense(64, activation='relu'))
model1.add(layers.Dense(10)) # 10 output classes for CIFAR-10
# Compile the model
model1.compile(optimizer='adam',loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True), metrics=['accuracy'])
```

```python
# Train the model
model1.fit(train_images, train_labels, epochs=10, validation_data=(test_images, test_labels))
```

```python
# Evaluate the model
test_loss, test_acc = model1.evaluate(test_images, test_labels, verbose=2)
print("\nTest accuracy:", test_acc)
```
