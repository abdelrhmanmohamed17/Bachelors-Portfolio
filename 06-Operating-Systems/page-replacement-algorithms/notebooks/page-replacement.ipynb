<font size="4">__FIFO Algorithm__</font>

```python
def fifo(page_references):
    
    """
    This function how work ?

    it takes page refrences as an argument as list

    every page passes through the loop

    and checks if the page is in the frames or not

    if it is in the frames then page faults is constant 

    else we update the page faults

    if the page already not in the frames
    
    we check if there is empty frame or not
    
    yes , we will append that page 
    
    no, we will remove the first page in the list then append that page
    
    After we end our loop we return the NO. of page faults and
    
    shape of last frame

    """
    frames = []
    page_faults = 0
    for page in page_references:
        if page not in frames:
            page_faults += 1
            if len(frames) < 3:
                frames.append(page)
            else:
                frames.pop(0)
                frames.append(page)
    return page_faults,frames
# Example

page_references = [1, 2, 3, 4, 1, 2, 5, 6, 1, 2, 3, 7, 8, 7, 8, 9, 7, 8, 9, 5]


faults_frame = fifo(page_references)
print("Total Page Faults and the last frame will be:", faults_frame)

```

<font size="4">__LFU Algorithm__</font>

```python
def lfu(page_references):
    """
    This function how work ?

    it takes page refrences as an argument as list

    every page passes through the loop

    and checks if the page is in the frames or not

    if it is in the frames then page faults is constant and update the fraquancy of that page

    else we update the page faults

    if the page already not in the frames
    
    we check if there is empty frame or not
    
    yes , we will append that page
    
    and update frequancy of that page to be 1 in the frequancy dictionary
    
    no, we will remove the page with lowest frequancy by
    
    checking the key (page) with the lowest frequancy in the dictionary
    
    then we remove it from the frames list and from the key from the frequancy dict.
    
    then add the page and make their frequancy value is 1
    
    After we end our loop we return the NO. of page faults and
    
    shape of last frame

    """
    frames = []
    page_faults = 0
    frequency = {}
    
    for page in page_references:
        if page not in frames:
            page_faults += 1
            if len(frames) < 3:
                frames.append(page)
                frequency[page] = 1
            else:
                # Frequancy lowest
                min_freq_page = min(frames, key=frequency.get)
                frames.remove(min_freq_page)
                del frequency[min_freq_page]
                frames.append(page)
                frequency[page] = 1
        else:
            frequency[page] += 1

    return page_faults,frames


# Example
page_references = [1, 2, 3, 4, 1, 2, 5, 6, 1, 2, 3, 7, 8, 7, 8, 9, 7, 8, 9, 5]
faults_frame = lfu(page_references)
print("Total Page Faults and the last frame will be:", faults_frame)

```

<font size="4">__MFU Algorithm__</font>

```python
def mfu(page_references):
    """
    This function how work ?

    Same as LFU we only remove the page with the highest frequancy

    """
    frames = []
    page_faults = 0
    frequency = {}
    
    for page in page_references:
        if page not in frames:
            page_faults += 1
            if len(frames) < 3:
                frames.append(page)
                frequency[page] = 1
            else:
                # Frequancy highest
                max_freq_page = max(frames, key=frequency.get)
                frames.remove(max_freq_page)
                del frequency[max_freq_page]
                frames.append(page)
                frequency[page] = 1
        else:
            frequency[page] += 1

    return page_faults,frames


# Example
page_references = [1, 2, 3, 4, 1, 2, 5, 6, 1, 2, 3, 7, 8, 7, 8, 9, 7, 8, 9, 5]
faults_frame = mfu(page_references)
print("Total Page Faults and the last frame will be:", faults_frame)
```

<font size="4">__Clock Algorithm__</font>

```python
def clock(page_references):
    """
    frames is a list that represents the frames, with -1 indicating an empty frame
    
    reference_bit is a list that keeps track of the reference bit for each frame, initialized to 0
    
    when a page fault occurs the algorithm scans through the frames using the pointer and checks the reference bit of each frame.
    
    if the reference bit is 0, the page is replaced in that frame and the reference bit is set to 1
    
    if the reference bit is 1, it is reset to 0, and the pointer moves to the next frame in a circular manner
    
    if the requested page is already in a frame, its reference bit is set to 1 to indicate it was recently referenced.
    """
    frames = [-1] * 3
    page_faults = 0
    pointer = 0
    reference_bit = [0] * 3

    for page in page_references:
        if page not in frames:
            page_faults += 1
            while True:
                if reference_bit[pointer] == 0:
                    frames[pointer] = page
                    reference_bit[pointer] = 1
                    pointer = (pointer + 1) % 3
                    break
                else:
                    reference_bit[pointer] = 0
                    pointer = (pointer + 1) % 3
        else:
            reference_bit[frames.index(page)] = 1
    return page_faults,frames
# Example
page_references = [1, 2, 3, 4, 1, 2, 5, 6, 1, 2, 3, 7, 8, 7, 8, 9, 7, 8, 9, 5]
faults_frame = clock(page_references)
print("Total Page Faults and the last frame will be:", faults_frame)
```
