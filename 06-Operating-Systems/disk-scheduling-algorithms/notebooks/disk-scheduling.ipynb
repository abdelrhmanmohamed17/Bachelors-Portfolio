## Scan

```python
def scan(start, requests, direction):
    # Sort requests in the order of their location on the disk
    requests.sort()

    # Initialize variables
    current_pos = start
    total_seek_time = 0
    n = len(requests)

    while requests:
        # Move the disk head in the current direction
        if direction == 'right':
            i = 0
            while i < n and requests[i] < current_pos:
                i += 1
            if i == n:
                direction = 'left'
                continue
            while i < n:
                current_request = requests.pop(i)
                total_seek_time += abs(current_request - current_pos)
                current_pos = current_request
                n -= 1
        else:
            i = n - 1
            while i >= 0 and requests[i] > current_pos:
                i -= 1
            if i == -1:
                direction = 'right'
                continue
            while i >= 0:
                current_request = requests.pop(i)
                total_seek_time += abs(current_request - current_pos)
                current_pos = current_request
                n -= 1

    return total_seek_time

```

## C-Scan

```python
def c_scan(start, requests, direction):
    requests.sort()
    current_pos = start
    total_seek_time = 0
    n = len(requests)
    while requests:
        # Move the disk head in the current direction
        if direction == 'right':
            i = 0
            while i < n and requests[i] < current_pos:
                i += 1
            if i == n:
                total_seek_time += abs(199 - current_pos)  # Move to track 199
                current_pos = 0  # Jump to track 0
                continue
            while i < n:
                current_request = requests.pop(i)
                total_seek_time += abs(current_request - current_pos)
                current_pos = current_request
                n -= 1
        else:
            i = n - 1
            while i >= 0 and requests[i] > current_pos:
                i -= 1
            if i == -1:
                total_seek_time += abs(0 - current_pos)  # Move to track 0
                current_pos = 199  # Jump to track 199
                continue
            while i >= 0:
                current_request = requests.pop(i)
                total_seek_time += abs(current_request - current_pos)
                current_pos = current_request
                n -= 1
    return total_seek_time
```

## F-Scan

```python
def f_scan(start, requests, direction):
    # Sort requests in the order of their location on the disk
    requests.sort()
    # Split requests into two separate queues
    before = [r for r in requests if r < start]
    after = [r for r in requests if r >= start]

    # Initialize variables
    current_pos = start
    total_seek_time = 0

    # Service requests in the first queue in the current direction
    while before:
        if direction == 'right':
            i = 0
            while i < len(before) and before[i] < current_pos:
                i += 1
            if i == len(before):
                total_seek_time += abs(199 - current_pos)  # Move to track 199
                current_pos = 0  # Jump to track 0
                continue
            while i < len(before):
                current_request = before.pop(i)
                total_seek_time += abs(current_request - current_pos)
                current_pos = current_request
        else:
            i = len(before) - 1
            while i >= 0 and before[i] > current_pos:
                i -= 1
            if i == -1:
                total_seek_time += abs(0 - current_pos)  # Move to track 0
                current_pos = 199  # Jump to track 199
                continue
            while i >= 0:
                current_request = before.pop(i)
                total_seek_time += abs(current_request - current_pos)
                current_pos = current_request
    # Jump to the other end of the disk and reverse the direction
    total_seek_time += 199
    current_pos = 199
    direction = 'left'
    # Service requests in the second queue in the reversed direction
    while after:
        if direction == 'right':
            i = 0
            while i < len(after) and after[i] < current_pos:
                i += 1
            if i == len(after):
                total_seek_time += abs(199 - current_pos)  # Move to track 199
                current_pos = 0  # Jump to track 0
                continue
            while i < len(after):
                current_request = after.pop(i)
                total_seek_time += abs(current_request - current_pos)
                current_pos = current_request
        else:
            i = len(after) - 1
            while i >= 0 and after[i] > current_pos:
                i -= 1
            if i == -1:
                total_seek_time += abs(0 - current_pos)  # Move to track 0
                current_pos = 199  # Jump to track 199
                continue
            while i >= 0:
                current_request = after.pop(i)
                total_seek_time += abs(current_request - current_pos)
                current_pos = current_request
    return total_seek_time
```
